# 🦠 COVID-19 Interactive Dashboard with Widgets
This Jupyter Notebook lets you explore COVID-19 data interactively by selecting a country. It uses `pandas`, `plotly`, and `ipywidgets`.

```python
# 📥 Load COVID-19 Dataset
import pandas as pd

url = "https://covid.ourworldindata.org/data/owid-covid-data.csv"
df = pd.read_csv(url, parse_dates=['date'])

# Preview the data
df[['location', 'date', 'total_cases', 'new_cases', 'total_deaths']].head()
```

```python
# 🧼 Prepare Country Dropdown
import ipywidgets as widgets
from IPython.display import display

countries = sorted(df['location'].dropna().unique())
country_dropdown = widgets.Dropdown(
    options=countries,
    value='India',
    description='Country:',
    style={'description_width': 'initial'}
)
display(country_dropdown)
```

```python
# 📊 Plotting Function
import plotly.express as px

def plot_covid_data(country):
    filtered = df[df['location'] == country]
    
    fig = px.line(
        filtered, 
        x='date', 
        y='total_cases', 
        title=f'Total COVID-19 Cases in {country}', 
        labels={'total_cases': 'Total Cases'},
        template='plotly_dark'
    )
    fig.show()
```

```python
# 🔄 Interactive Widget
from ipywidgets import interact
interact(plot_covid_data, country=country_dropdown);
```
